```python
import time
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
search = input("Enter model to search: ")
url = f"https://ollama.com/search?q={search}"
```

```python
page = requests.get(url)
print(page)
```

```python
soup = BeautifulSoup(page.content,'html.parser')
# print(soup)
```

```python
search_results = soup.find_all('span',attrs={'x-test-search-response-title':True})
print(len(search_results))

links = soup.find_all('a',attrs={'class':"group w-full",'href':True})

model_links = []

for link in links:
    print(link['href'])
    model_links.append(f"https://ollama.com{link['href']}")

model_names = []
for result in search_results:
    print(result.text)
    model_names.append(result.text)
```

```python
print(model_names)
print(model_links)
```

```python
df = pd.DataFrame({'Model':model_names,'Link':model_links})
print(df)
```

```python
df.to_csv(f'{search}_models_list.csv')
```
